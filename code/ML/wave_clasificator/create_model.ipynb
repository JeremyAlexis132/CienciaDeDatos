```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.cluster import DBSCAN
from sklearn.cluster import KMeans

env = 'project'
```

```python
df_model = (
    spark.sql(
        f"""
            SELECT coast_name, datetime, wind_u, wind_v, wave_u, wave_v, wave_period_s
            FROM cor_{env}.silver.swell_metrics
        """
    )
).toPandas()
```

```python
features = ['wind_u', 'wind_v', 'wave_u', 'wave_v', 'wave_period_s']

pipeline_dbscan = Pipeline([
    ('scaler', StandardScaler()),
    ('dbscan', DBSCAN(eps=0.5, min_samples=5))
])

pipeline_dbscan.fit(df_model[features])
df_model['dbscan_label'] = pipeline_dbscan.named_steps['dbscan'].labels_
```
